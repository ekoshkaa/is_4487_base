```python
import matplotlib.pyplot as plt
```

```python
processed = results_df["LLM_Output"].str.split("|", expand=True)
processed.columns = ["Sentiment", "Theme", "Word Count", "Suggested Action", "Character Count", "Star Score"]

processed["Character Count"] = processed["Character Count"].astype(int)
processed["Star Score"] = processed["Star Score"].astype(int)
```

```python
processed["Theme"].value_counts().plot(kind="bar")
plt.title("Most Common Themes")
plt.xlabel("Theme")
plt.ylabel("Count")
plt.show()
```

```python
processed["Sentiment"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Sentiment Distribution")
plt.ylabel("")
plt.show()
```

```python
pd.crosstab(processed["Theme"], processed["Sentiment"]).plot(kind="bar", stacked=True)
plt.title("Sentiment by Theme")
plt.xlabel("Theme")
plt.ylabel("Count")
plt.show()
```

```python
processed["Character Count"].plot(kind="hist")
plt.title("Distribution of Review Length (Characters)")
plt.xlabel("Characters")
plt.show()
```

### ✍️ Your Response: 🔧
1. I chose a bar chart for theme frequency because it clearly shows the main customer concerns.
A pie chart works well for showing the overall sentiment breakdown at a high level.
A stacked bar chart shows how sentiment varies for each theme, helping identify which themes have the highest negative feedback.
A histogram is best for visualizing numeric values like character count. It shows whether longer reviews are associated with stronger emotions.
